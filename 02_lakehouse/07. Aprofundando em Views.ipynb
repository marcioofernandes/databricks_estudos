### O que é uma View? 

Uma view é uma consulta SQL armazenada que funciona como uma "tabela virtual". Ela não armazena dados fisicamente (salvo materialized views) apenas armazena a definição da consulta. Quando você acessa a view, a consulta é executada em tempo por debaixo dos panos. 

Imagina que tem 3 tabelas e você precisa de um resultado do join dessas 3 tabelas, nisso é criado a consulta(query) onde você fará a lógica que precisa para obter os dados que necessita. Mas escrever sempre essa lógica, é trabalhoso, e é ai que as views entram. Uma View nada mais é que o resultado dessa consulta, então sempre que você faz um select * em uma view, você está rodando a consulta por trás. 

A maior vantagem disso é que você não ocupa espaço de armazenamento, mas a desvantagem é que você gasta no processamento sempre que faz uma consulta na view. 

### Tipos de Views

#### Standard View (Padrão)

É a View padrão, os dados não são gravados no armazenamento, apenas os metadados, ou seja, este tipo view fica persistido no metastore, seus dados como nome de colunas, schema, query, tudo isso fica armazenado no metastore, e mesmo quando você desliga a sessão, ela permece lá e  toda vez que uma view é consultada, o mecanismo de processamento precisa executar a consulta e ler os dados das tabelas base para calcular os resultados no momento. 

- A DEFINIÇÃO DA QUERY é armazenada no metastore
- Os DADOS NÃO são armazenados (não ocupa espaço de storage)
- A view continua existindo mesmo após encerrar a sessão
- Múltiplos usuários podem acessá-la
- Mas os dados são calculados toda vez que você acessa

**Quando usar:** Use para simplificar consultas complexas, encapsular lógicas de negócios e fornecer uma interface de consumo consistente para os usuários, evitando a duplicação dos dados no armazenamento

#### Temporary Views 

As views temporárias são views de exibição efêmeras, ou seja, ela só existe durante o tempo de execução de uma sessão ou de um pipeline de dados (Como por exemplo o Lakeflow Spark Declarative Pipelines). Elas são privadas e nunca são publicadas no catalog. 

Aprofundando os detalhes sobre o funcionamento e o propósito das temporary views no Databricks (especialmente dentro de pipelines como o Lakeflow Spark Declarative Pipelines), temos os seguintes pontos:

1. Ciclo de Vida e Privacidade Restrita Diferente das tabelas ou views tradicionais, uma temporary view existe apenas durante o tempo de execução (ciclo de vida) do pipeline em que foi definida

- Ela é um objeto estritamente privado daquele pipeline específico e não é publicada ou adicionada a catálogos públicos (como o Unity Catalog)
- Um comportamento interessante é que uma temporary view pode ter exatamente o mesmo nome de uma tabela ou view oficial que já existe no catálogo
- Quando isso acontece, qualquer consulta feita dentro daquele pipeline usando esse nome será roteada (resolvida) para a temporary view local, ignorando temporariamente o objeto do catálogo público

2. Principais Casos de Uso (Por que usá-las?) As temporary views são desenhadas para resolver problemas de arquitetura do pipeline sem gerar custos desnecessários:

- Simplificação de consultas e etapas intermediárias: Se você possui uma transformação muito longa ou complexa, pode quebrá-la em queries menores usando temporary views. Elas servem como passos de preparação para os dados antes de serem finalmente gravados em uma tabela persistente (streaming table ou materialized view)

- Reutilização dentro do pipeline: Se uma mesma transformação inicial vai alimentar várias tabelas diferentes mais adiante no seu fluxo, você cria uma temporary view, isso garante que o sistema compreenda as dependências visuais no grafo do pipeline, permitindo que você referencie essa view pelo nome em várias consultas subsequentes

- Redução de custos: Como elas apenas processam em memória e não materializam dados no armazenamento em nuvem, você evita custos extras de processamento e disco com tabelas intermediárias que não têm utilidade de negócios para os usuários finais

- Verificações de Qualidade de Dados (Expectations): É muito comum usar temporary views para aplicar regras de qualidade de dados ou construir padrões de "quarentena". Você cria uma temporary view que lê os dados brutos, aplica regras que identificam registros válidos ou inválidos, e então as tabelas de destino finais consultam essa view filtrando apenas o que está correto

3. Como declará-las (Sintaxe) Você pode criar temporary views tanto usando SQL quanto Python:
Em Python: Utiliza-se o decorador @dp.temporary_view() (disponibilizado importando o módulo pyspark.pipelines). Isso substitui o formato mais antigo, que usava o decorador @dlt.view no antigo Delta Live Tables

- Em SQL: O padrão recomendado é o comando CREATE TEMPORARY VIEW

- Historicamente, eram usados os comandos CREATE LIVE VIEW e CREATE TEMPORARY LIVE VIEW, a única ressalva importante feita pela documentação é que, caso você queira incluir regras de qualidade de dados (expectations) diretamente usando cláusulas CONSTRAINT no SQL, você ainda deve usar a sintaxe CREATE LIVE VIEW, pois a mais nova não suporta cláusulas constraint nativamente

#### Materialized View

Uma Materialized View é um objeto declarativo de pipeline que armazena fisicamente (em cache) os resultados pré-calculados de uma consulta SQL. Diferente das views padrão, que precisam executar a consulta do zero lendo as tabelas base toda vez que são acessadas, as materialized views entregam o dado já processado, o que reduz drasticamente a latência da consulta e o consumo de recursos computacionais. 

Aqui estão os detalhes mais importantes sobre como elas funcionam e quando utilizá-las:

1. **Atualizações (Refreshes) Inteligentes e Incrementais:** O grande diferencial da materialized view é o seu motor de atualização subjacente (o Enzyme), que mantém a tabela sincronizada com as fontes de dados de origem, isso quer dizer que o sistema analisa a consulta e escolhe automaticamente entre dois tipos de atualização baseando-se no que for mais barato e eficiente

- Refresh Incremental: Sempre que possível, o Databricks processa apenas os dados novos ou alterados desde a última atualização e os anexa na view. Para garantir que o refresh incremental funcione, as tabelas de origem (em formato Delta) devem estar com as propriedades de rastreamento de linha (row tracking) e Change Data Feed (CDF) habilitadas. 

- Refresh Completo (Full Recompute, ou full refresh): O sistema limpa a tabela inteira e reprocessa todos os dados do zero. Isso ocorre quando o algoritmo detecta mudanças muito severas na origem (como dados removidos), ou quando a consulta usa lógica que impede a incrementalidade, como funções não determinísticas (ex: RANDOM(), CURRENT_TIMESTAMP())

Se necessário, você pode controlar esse comportamento através da cláusula REFRESH POLICY, podendo forçar a atualização para ser AUTO (padrão), INCREMENTAL, INCREMENTAL STRICT (que faz o processo falhar caso a consulta não suporte incrementalidade) ou FULL

2. **Gatilhos e Agendamentos** Ao invés de depender de orquestradores externos complexos para saber a hora certa de atualizar os dados de uma view, as materialized views suportam controle de execução nativo

- **TRIGGER ON UPDATE:** A view monitora as tabelas de origem e se atualiza automaticamente assim que mudanças nessas fontes são detectadas. Isso mantém o dado atualizado sem requerer lógicas de agendamento complexas

- **SCHEDULE:** Você pode estipular intervalos de tempo (como SCHEDULE EVERY 1 HOUR ou usar expressões Cron) caso possua SLAs de dados ou cadências de atualização estritas no seu negócio

3. **Casos de Uso na Arquitetura Medallion** Devido à sua semântica de processamento em lote (batch), elas garantem que os resultados serão sempre corretos mesmo que os dados cheguem atrasados ou fora de ordem. Sendo assim, são a escolha perfeita para a camada Ouro (Gold) do seu Data Lakehouse. Elas são amplamente usadas para: 

- Alimentar painéis de Business Intelligence (BI) para os usuários finais (Dashboards)

- Computar desnormalizações, grandes junções (joins) entre tabelas e agregações complexas (como contagens e somas)

- Absorver lógicas que envolvam atualizações e deleções em massa nativamente (ao contrário das Streaming Tables, que são restritas ao modo de adição contínua de novos registros)

4. **Como declará-las** No ecossistema do Lakeflow Spark Declarative Pipelines, você pode defini-las da seguinte maneira:

- Via SQL: Utilizando o comando CREATE OR REFRESH MATERIALIZED VIEW nome_da_view AS SELECT ...

- Via Python: Utilizando o decorador @dp.materialized_view() imediatamente acima de uma função que retorne o seu DataFrame processado

**Principais Limitações a Considerar**

- Sem Viagem no Tempo: Diferente das tabelas Delta comuns ou streaming tables, elas não suportam consultas de Time Travel (acesso a versões do histórico)

- Não são para streaming de curtíssima latência: A latência de atualização e propagação do dado costuma ser na casa dos segundos ou minutos, e não milissegundos

- Identificadores: Elas não suportam a criação de colunas de identidade ou chaves substitutas (surrogate keys)

```sql
-- Criando uma view Ad-hoc (neste caso, para atualizar essa view é necessário rodar o comando de Refresh)
CREATE OR REPLACE MATERIALIZED VIEW total_vendas AS 
select 
  data_da_compra, 
  sum(valor_compra) as total_vendas 
from vendas 
group by data_da_compra; 

-- Não foi possível criar a materialized view por conta da conta gratuita não ter acesso ao Serverless Generic Compute, porém o comando é esse. 
```

```sql
REFRESH MATERIALIZED VIEW total_vendas; -- Para atualizar a view
```

```sql
DROP MATERIALIZED VIEW total_vendas; -- Para deletar a view
```

```sql
-- Adicionando Trigger na view, assim sempre que a tabela receber atualização ela atualiza também

CREATE OR REPLACE MATERIALIZED VIEW total_vendas TRIGGER ON UPDATE
AS 
select 
  data_da_compra, 
  sum(valor_compra) as total_vendas 
from vendas 
group by data_da_compra; 
```

```sql
-- Criando materilized View com Schedule 
CREATE OR REPLACE MATERIALIZED VIEW total_vendas
  SCHEDULE CRON '0 0 10 * * ?' AT TIME ZONE 'UTC'
AS 
select 
  data_da_compra, 
  sum(valor_compra) as total_vendas 
from vendas 
group by data_da_compra; 
```

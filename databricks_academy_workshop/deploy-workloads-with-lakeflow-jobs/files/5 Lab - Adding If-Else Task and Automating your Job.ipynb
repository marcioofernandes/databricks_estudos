## E. Agendando Seu Job Usando a UI de Jobs

Siga estes passos para agendar seu job no Databricks:

1. **Abra Seu Job:** Vá para o job que você acabou de criar.

2. **Vá para a Aba de Tasks:** Certifique-se de estar visualizando a aba **Tasks** dentro do seu job.

3. **Expanda os Detalhes do Job:** No lado direito da UI de Jobs, encontre o painel **Job Details**.  
   - **NOTA:** Se o painel estiver recolhido, clique no ícone de seta para expandi-lo.

4. **Adicione um Agendamento:** Na seção **Schedules & Triggers**, clique em **Add trigger**. Você verá três opções de agendamento:  
   - **Scheduled** (executa em horários específicos)

   - **Continuous** (executa assim que a execução anterior termina)

   - **Table Update** (executa automaticamente sempre que uma ou mais tabelas especificadas são atualizadas)

   - **File arrival** (executa quando arquivos chegam em um local)

5. **Configure uma Execução Agendada:**  
   - Escolha **Scheduled**.

   - Clique na seção **Advanced** para ver mais opções.

6. **Configure o Agendamento:**  
   - Defina o job para executar **todos os dias** em um horário de sua escolha.

   - Certifique-se de selecionar seu fuso horário específico.

   - **Dica:** Defina o agendamento para começar dois minutos a partir do horário atual para não precisar esperar muito para o job rodar.

**NOTA:** Agendar seu job garante que ele execute automaticamente nos horários especificados. Você também pode iniciar a execução do job clicando em **Run Now** no topo do seu job.

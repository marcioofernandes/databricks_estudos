```sql
select * from read_files(
    "dbfs:/databricks-datasets/bikeSharing/data-001/*.csv",
    format => "csv", 
    header => true, 
    inferschema => true
)
limit 5; 
```

```sql
-- Criando tabelas externas (CREATE TABLE USING) com arquivos CSVs 
-- Novamente esse comando não funciona usando pastas dbfs no UC, sendo necessário um path externo 
-- A Célula fica de exemplo de construção


CREATE TABLE bike_sharing 
USING CSV
OPTIONS(
    PATH "dbfs:/databricks-datasets/bikeSharing/data-001/*.csv", 
    header 'true', 
    inferschema 'true',
    sep ','
)
```

Outra informação relevante é que o inferschema as vezes não funciona direito com arquivos .csv, para um resultado mais assertivo, desclarar o schema como mostra na célula abaixo. 

```sql
-- criar com schema manual do seu jeito 
CREATE TABLE bike_sharing_tratado (
    instant STRING,
    dteday DATE,
    season STRING,
    yr INT,
    mnth INT,
    hr INT,
    holiday STRING,
    weekday STRING,
    workingday STRING,
    weathersit STRING,
    temp STRING,
    atemp STRING,
    hum STRING,
    windspeed STRING,
    casual STRING,
    registered STRING,
    cnt STRING
)
USING CSV
OPTIONS (
    path '<path_do_bucket>',
    header 'true',
    sep ','
);
```

```sql
select * from read_files(
    "dbfs:/databricks-datasets/bikeSharing/data-001/*.csv", 
    format => 'csv',
    header => true, 
    inferschema => true
) limit 5 ;
```

```sql
-- Criando Tabela no formato Delta com  (CTAS)
CREATE TABLE bike_sharing_delta AS
select * from read_files(
    "dbfs:/databricks-datasets/bikeSharing/data-001/*.csv", 
    format => 'csv',
    header => true, 
    inferschema => true
)
```
